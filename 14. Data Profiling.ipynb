###Data profiling in Databricks
  1. Profile data using UI.
  2. Profile data using DBUTILS Package
  3. Profile data manually:
    1. Count
    2. Countif
    3. Min
    4. Max
    5. Where clause

###1. Profile data using User interface.

```python
%sql
select * from ginzobox.bronze.v_customers;
```

###2. Profile Data using DBUTILS Package

```python
%python
df= spark.table('ginzobox.bronze.v_customers')
dbutils.data.summarize(df)
```

###3. Profile Data Manually

```python
%sql
select count(*),count(email) from ginzobox.bronze.v_customers;
```

```python
%sql
select count_if(customer_id is null) from ginzobox.bronze.v_customers;
```

```python
%sql
select count(*) from ginzobox.bronze.v_customers where customer_id is null;
```
