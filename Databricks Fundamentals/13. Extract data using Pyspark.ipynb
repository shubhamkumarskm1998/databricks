###Extract Data using Pyspark
  1. run sql commands using sql spark.sql function.
  2. Spark Dataframe reader api
  3. read tables using spark.table function

###1. Run Sql commands using spark.sql

```python
df=spark.sql("""select * from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers/customers_2024_10.json`""")
df.display()
```

###2. Spark reader api 

```python

df= spark.read.format('json').load('dbfs:/Volumes/ginzobox/landing/operational_data/customers/customers_2024_10.json')
df.display()
```

###3. Spark.table

```python
df_payments=spark.table('ginzobox.bronze.v_payments')
df_payments.display()
```
