###Extract Data from customers json file
  1. Query single file
  2. Query list of files using wildcard characters.
  3. Query all the files in the folder.
  4. Select file metadata
  5. Register files in unity catalog using views.
  6. Temporary views
  7. Global Temporary views.

####1. Query Single file

```sql
Use catalog ginzobox;
use schema landing;
```

```sql
%fs ls /Volumes/ginzobox/landing/operational_data/customers/
```

```sql
select * from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers/customers_2024_10.json`
```

###2. Query multiple files using wildcard characters

```sql
select * from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers/customers_2024_*.json`
```

###3. Query all the files from the folder

```sql
select * from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers/`
```

###4. Select file metadata

```sql
select input_file_name() as file_path,
_metadata.file_path as file_path,
_metadata.file_name as file_name,
_metadata.file_size as file_size,
_metadata.file_block_start as file_block_start,
_metadata.file_block_length as file_block_length,
* from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers/`
```

###5. Register files in unity catalog as views.

```sql
Create or replace view ginzobox.bronze.v_customers as 
select *, _metadata.file_path as file_path from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers`
```

```sql
select * from ginzobox.bronze.v_customers;
```

###6. Temporary Views
a new spark session starts when we start the notebook so when you detach the notebook with cluster, the temporary view gets destroyed.

```sql
create or replace temporary view ginzobox.bronze.v_temp_customers as select *, _metadata.file_path as file_path from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers`
```

###7. Global Temporary Views 
it ends when spark application ends as compared to temporary views which ends when spark session is ended.

```sql
create or replace global temporary view ginzobox.bronze.v_global_temp_customers as select *, _metadata.file_path as file_path from JSON.`dbfs:/Volumes/ginzobox/landing/operational_data/customers`
```
