```sql
Create table if not exists ginzobox.bronze.v_payments
(payment_id integer,
order_id integer,
payment_timestamp timestamp,
payment_status integer,
payment_method string)
using csv
options (
 header=true,
 delimiter=','
)
Location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/landing/external_data/payments/'
```
