### Databricks Intelligence Platform
- A Data Lakehouse combines Data Lake scalability with Data Warehouse reliability (ACID, schema, performance) for all data types.
`Example: Store raw + processed data in Delta tables → run BI queries + ML on same data.
Methods:
    - Data Lake → Cheap storage for all data types (raw, unstructured).
    - Data Warehouse → Structured, optimized for BI queries.
    - Lakehouse → Combines both (Delta Lake adds ACID, schema, performance).`
- Use Auto-Stop to automatically shut down a SQL warehouse after inactivity.
`Example: Set Auto-Stop = 10 mins → warehouse stops if no queries run for 10 minutes.
Methods:
    - Auto-Stop → Stops SQL warehouse after idle time (cost saving).
    - Auto Termination → For clusters (not SQL warehouses).
    - Scaling Policy → Adjusts compute size, not shutdown.`
- Databricks uses a control plane (managed by Databricks) and data plane (runs in customer cloud) to keep data within the customer environment.
`Example: Notebooks + jobs managed by Databricks (control plane) → execution + data access happen in customer’s cloud (data plane).
Methods:
    - Control Plane → UI, job scheduler, APIs (Databricks-managed).
    - Data Plane → Compute clusters + data processing (customer cloud).
    - Benefit → Data never leaves customer environment (security/compliance).`
- All-purpose clusters are persistent for interactive use, while job clusters are ephemeral and run only for a job.
`Example: Dev team uses all-purpose cluster for notebooks → scheduled ETL uses job cluster that auto-terminates after run.
Methods:
    - All-purpose cluster → Interactive, shared, long-running.
    - Job cluster → Ephemeral, per-job, auto-terminates.
    - Use case → Dev/analysis vs production pipelines.`
- Use a shared cluster with Databricks Runtime for Machine Learning to ensure consistent environments and pre-installed ML libraries.
`Example: All team members attach to ML Runtime cluster → same Spark, Python, MLflow, scikit-learn versions.
Methods:
    - Databricks Runtime for ML → Pre-installed ML libraries + consistent environment.
    - Shared cluster → Ensures all users run same config.
    - Manual installs → Inconsistent (avoid).`
- **Databricks notebooks support live co-editing, so teams can collaborate on the same notebook at the same time—no branch juggling needed.**
- **Schedule the query to refresh directly from the Queries page in Databricks SQL for automated daily updates.**
- **Cluster pools speed up job or notebook startup by keeping ready-to-go instances, reducing wait time for clusters.**
- **You can version control job configurations by downloading the job’s JSON definition and storing it in your Git repo**
- **Fact table → granular transactional data (events, metrics)**
- **Dimension table → descriptive attributes (customer, product, etc.)**
- **Triggered + Development Mode = Single run, pipelne stops and compute stays alive**
- **Triggered + Production = single run, pipeline stops and compute is terminated**

### Development and Ingestion
- Use Databricks Connect to develop and debug in IDE while executing Spark on a remote cluster.
`Example: Write/debug code in PyCharm → connect via Databricks Connect → run on cluster with full dataset.`
- In Databricks Connect, Spark code runs on the cluster while regular Python code runs locally in the IDE.
`Example: df.collect() → Spark executes on cluster, then list comprehension runs locally on collected data.
Methods:
    - Spark transformations/actions → Run on Databricks cluster (distributed).
    - Python logic (loops, list comprehension) → Runs locally.
    - collect() / toPandas() → Brings data from cluster → local execution.`
- Use input_file_name() to get the source file path for each row.
``Example: SELECT *, input_file_name() FROM json.\/Volumes/retail/landing/customers/`;`
Methods:
    - input_file_name() → Returns full file path for each row (most common).
    - _metadata.file_path → Newer approach (Spark 3.4+), similar purpose.
    - get_json_object → Extracts JSON fields (not for file path).``
- Using cloudFiles with readStream indicates Auto Loader for streaming file ingestion from cloud storage.
Example: spark.readStream.format("cloudFiles").option("cloudFiles.format","json").load(path) → incrementally ingests new JSON files.
Methods:
    - Auto Loader (cloudFiles) → Incremental file ingestion (recommended).
    - readStream → Enables streaming ingestion.
    - COPY INTO → Batch file ingestion (not streaming).
    - Kafka source → .format("kafka") for event streams.
- Use magic commands (%python, %sql, %scala, %sh) to run multiple languages in the same notebook.
`Example: %sql SELECT * FROM table; then %python df.show() in next cell.
Methods:
    - %python → Run Python code
    - %sql → Run SQL queries
    - %scala → Run Scala code
    - %sh → Run shell commands`
- Use json.\path`` syntax to read JSON files directly in Spark SQL without a table.
`Example: SELECT * FROM json.\/Volumes/data/customers/`;`
Methods:
    - json.\path`` → Read JSON files directly via SQL.
    - parquet.\path`` → Read Parquet files similarly.
    - spark.read.json() → PySpark equivalent (not SQL).`
- Databricks Repos supports basic Git operations like pull, commit, and push directly from the UI.
`Example: Pull latest changes → edit notebook → commit → push to remote repo.
Methods:
    - Pull → Sync latest changes from remote.
    - Commit → Save changes locally with message.
    - Push → Upload changes to remote repo.
    - PRs/CI/CD → Handled outside (GitHub, GitLab, etc.).`
- Databricks Connect requires the local Python version to match the cluster runtime (e.g., Python 3.11).
`Example: Cluster = Python 3.11 → local must also be Python 3.11, otherwise session fails.
Methods:
    - Version match → Local Python = Cluster Python (mandatory).
    - Databricks Connect → Bridges IDE ↔ cluster execution.
    - Mismatch issue → Causes runtime/session startup errors.`
- Create a View to provide reusable, governed access without storing physical data.
``Example: CREATE VIEW customer_view AS SELECT * FROM json.\/path`;`
Methods:
    - View → Logical layer (no data storage, reusable, governed).
    - Managed/External Table → Stores physical data (not required here).
    - Temp Views → Session-scoped (not persistent).``
- Auto Loader defaults to STRING types for JSON/CSV because these formats don’t include schema information.
`Example: Add .option("cloudFiles.inferColumnTypes","true") → infers INT, BOOLEAN, etc.
Methods:
    - Default behavior → All columns as STRING (safe inference).
    - inferColumnTypes = true → Enables type inference.
    - Explicit schema → Best practice for production.`
- Dropping an external table removes only metadata, not the underlying data files.
`Example: DROP TABLE bronze.products; → table removed, files in storage remain intact.
Methods:
    - External table → Metadata deleted, data preserved.
    - Managed table → Metadata + data both deleted.
    - LOCATION clause → Indicates external table.`

### Data Processing & Transformations

- Use transform_values to update map values (e.g., apply 20% tax) while keeping keys unchanged.
`Example: SELECT transform_values(item_prices, (k, v) -> round(v * 1.20, 2)) FROM order_items;`
- INSERT OVERWRITE TABLE to fully refresh a Delta table while preserving schema and metadata.
`Example: INSERT OVERWRITE TABLE customer_data SELECT * FROM staging_table;
Methods:
    - INSERT OVERWRITE → Full data replacement (best for complete refresh).
    - MERGE INTO → Incremental updates (upsert based on conditions).
    - DELETE + INSERT → Two-step full refresh (less efficient, non-atomic).
    - CREATE OR REPLACE TABLE → Recreates table (may lose metadata, not preferred).`
- Use OPTIMIZE to compact small files into larger ones and improve read performance.
`Example: OPTIMIZE table_name;
Methods:
    - OPTIMIZE → Compacts small files (improves read performance).
    - OPTIMIZE ZORDER BY → Compaction + data clustering for faster filtering.
    - VACUUM → Removes old/unreferenced files (storage cleanup, not compaction).
    - ANALYZE TABLE → Updates stats for query planning (no file impact).`
- ACID in Delta Lake ensures reliable transactions (no partial writes + consistent reads during concurrency).
`Example: If a write job fails midway → no partial data is saved; readers still see last consistent snapshot.
Methods:
    - Atomicity → All-or-nothing writes (commit or rollback).
    - Consistency → Data remains valid after transaction.
    - Isolation → Concurrent operations don’t conflict (snapshot view).
    - Durability → Committed data is permanently stored.`
- Use flatten() to convert an array of arrays into a single flat array.
`Example: SELECT flatten(product_categories) FROM table_name;
Methods:
    - flatten() → Flattens nested arrays into one array.
    - explode() → Converts array elements into multiple rows (not flattening).
    - flat_map → PySpark only (not SQL).`
- Use Trigger.ProcessingTime("1 minute") to run micro-batches every 60 seconds.
`Example: .trigger(Trigger.ProcessingTime("1 minute")) → processes new data every 60 sec.
Methods:
    - ProcessingTime("1 minute") → Scheduled micro-batch (recurring).
    - Trigger.Once() → Runs once and stops.
    - Trigger.AvailableNow() → Processes all available data then stops.
    - Trigger.Continuous() → Continuous processing (not micro-batch).`
- Use %run to include and reuse shared notebook code across multiple notebooks.
`Example: %run /Shared/Includes/config → loads configs, functions, and variables into current notebook.
Methods:
    - %run → Import another notebook’s code (recommended).
    - Copy-paste → Not maintainable (avoid).
    - Repos (modules) → Advanced reuse using Python files.`
- Use readStream → transform → writeStream with checkpointing for Bronze-to-Silver streaming.

Example:

`-    spark.readStream
 -               .format("delta")
 -               .load("/mnt/bronze")
 -               .filter("event_type IS NOT NULL")
 -               .select("user_id","event_type","timestamp")
 -               .writeStream.format("delta")
 -               .option("checkpointLocation","/mnt/checkpoints/silver")
 -               .start("/mnt/silver")

Methods:
    - readStream → Streaming read from Bronze.
    - filter + select → Apply transformations.
    - writeStream → Continuous write to Silver.
    - checkpointLocation → Fault tolerance & exactly-once processing.`
 - Use COUNT(*) WHERE column IS NULL or COUNT_IF(column IS NULL) to count NULL values.
`Example:
    - SELECT COUNT(*) FROM returns_data WHERE return_reason IS NULL;
    - SELECT COUNT_IF(return_reason IS NULL) FROM returns_data;
Methods:
    - COUNT(*) + WHERE IS NULL → Standard SQL approach.
    - COUNT_IF(condition) → Direct and cleaner in Databricks SQL.
    - COUNT(column) → Ignores NULLs (not suitable here).`
 - Use Triggered mode to run DLT pipelines periodically and stop compute between runs for cost savings.
`Example: Schedule pipeline every 4 hours → runs → processes data → shuts down cluster.
Methods:
    - Triggered mode → Batch-style runs, cost-efficient (best fit).
    - Continuous mode → Always running, low-latency but higher cost.
    - Auto Loader → Ingestion method (not a pipeline mode).`
 - A table created without a LOCATION is a managed table stored in the default metastore location.
`Example: CREATE TABLE bronze.sales (...) → data stored in managed storage → dropped table deletes data.
Methods:
    - Managed table → No LOCATION, Databricks manages data + metadata.
    - External table → Uses LOCATION, data not deleted on drop.
    - Temporary view → No data storage, session-scoped.`
 - Use schema_of_json() to infer the schema from a JSON string.
`Example: SELECT schema_of_json('{"id":1,"name":"abc"}');
Methods:
    - schema_of_json() → Infers schema (returns struct format).
    - from_json() → Parses JSON using schema.
    - get_json_object() → Extracts fields (not schema).`
 - Use transform() to apply a function to each array element while preserving structure.
`Example: SELECT transform(items, x -> upper(x)) FROM table_name;
Methods:
    - transform() → Modify each element in array (keep structure).
    - explode() → Converts array to rows (changes structure).
    - filter() → Removes elements based on condition.
    - flatten() → Flattens nested arrays.`
 - Use Medallion layers → Bronze (raw JSON), Silver (cleaned/standardized), Gold (aggregated for analytics).
Methods:
    - Bronze → Raw ingestion (no transformations).
    - Silver → Data cleaning, schema enforcement, enrichment.
    - Gold → Aggregations for BI/ML (KPIs, trends).
- Use a Temporary View to reuse query results within a session without storing data permanently.
`Example:
CREATE OR REPLACE TEMP VIEW temp_clicks AS SELECT * FROM clicks WHERE ...;
Methods:
    - Temporary View → Session-scoped, no storage, auto-deleted (best fit).
    - Global Temp View → Shared across sessions (persists while cluster runs).
    - Managed/External Tables → Persist data in storage (not suitable).`
- Use CREATE OR REPLACE FUNCTION with typed parameters and RETURN expression for SQL UDFs.
`Example:
CREATE OR REPLACE FUNCTION format_name(fname STRING, lname STRING) RETURNS STRING RETURN INITCAP(fname) || ' ' || INITCAP(lname);
Methods:
    - CREATE OR REPLACE FUNCTION → Define reusable SQL logic.
    - INITCAP() → Capitalizes first letter of words.
    - || → String concatenation in Spark SQL.`
- Use spark.sql() to execute SQL queries on temporary views and return a DataFrame.
`Example: spark.sql("SELECT * FROM transactions_view WHERE amount > 1000")
Methods:
    - spark.sql() → Run SQL on views/tables (returns DataFrame).
    - createOrReplaceTempView() → Register DataFrame as SQL view.
    - spark.read → For reading data sources (not SQL execution).`
- Use a SQL UDF to centralize and reuse status code-to-text mapping across queries.
`Example:
CREATE FUNCTION map_status(code INT) RETURNS STRING RETURN CASE WHEN code=1 THEN 'Success' WHEN code=2 THEN 'Pending' END;
Methods:
    - SQL UDF → Reusable, centralized logic (best fit).
    - CASE WHEN → Repeated logic (not scalable).
    - Lookup table JOIN → Good for large mappings but heavier.`
- Use CREATE OR REFRESH MATERIALIZED VIEW for Gold layer aggregations with automatic, incremental refresh.
Example:
- CREATE OR REFRESH MATERIALIZED VIEW
- gold_metrics AS SELECT date, region, COUNT(DISTINCT user_id) AS dau,
- COUNT(*) AS clicks FROM LIVE.silver_clicks GROUP BY date, region;
Methods:
    - MATERIALIZED VIEW → Auto-refresh, best for Gold aggregations.
    - STREAMING TABLE → Real-time processing (Bronze/Silver).
    - Regular TABLE → Static, no auto-refresh.
- Use EXPECT ... ON VIOLATION DROP ROW to discard invalid rows, continue processing, and track metrics.
Example:
- CREATE OR REFRESH STREAMING TABLE clean_customers (
- CONSTRAINT valid_email EXPECT (email IS NOT NULL) ON VIOLATION DROP ROW ) AS SELECT * FROM STREAM(LIVE.raw_customers);
Methods:
    - DROP ROW → Removes invalid rows + logs metrics (best fit). expect_or_fail
    - WARN (default) → Keeps rows, logs violations. expect
    - FAIL UPDATE → Fails pipeline on violation. expect or fail
- Use PIVOT to convert data from long format to wide format (rows → columns).
Example:
SELECT * FROM monthly_sales PIVOT (SUM(sales_amount) FOR month IN ('Jan','Feb','Mar'));
Methods:
    - PIVOT → Converts rows into columns (long → wide).
    - GROUP BY → Aggregates data but doesn’t reshape structure.
    - EXPLODE → Expands data (opposite of pivot).

### Data Governance & Quality
- Alerts in Databricks SQL trigger notifications when a scheduled query result meets a defined condition.
`Example: If daily_sales < 10000 → alert sends email/webhook notification.
Methods:
    - SQL Alert → Monitors query results and triggers notifications.
    - Scheduled Query → Runs periodically to evaluate condition.
    - Notification channels → Email, webhook for external systems.`
- Grant USE CATALOG on the catalog to allow access through the Unity Catalog hierarchy.
`Example: GRANT USE CATALOG ON CATALOG analytics TO reporting_team;
Methods:
    - USE CATALOG → Access to catalog (top-level required).
    - USE SCHEMA → Access to schema within catalog.
    - SELECT → Read access on table.
    - Hierarchy rule → Need all 3 (catalog → schema → table).`
- DLT EXPECT logs violations but still writes invalid rows to the target table.
`Example: Negative/null age → row is written, violation tracked in DLT metrics.
Methods:
    - EXPECT (default) → Warn (log + allow data).
    - EXPECT ... ON VIOLATION DROP ROW → Drops invalid rows.
    - EXPECT ... ON VIOLATION FAIL UPDATE → Fails pipeline on violation.`
- Use DESCRIBE DETAIL or DESCRIBE EXTENDED to find a table’s storage location.
`Example:
    - DESCRIBE DETAIL sales.orders;
    - DESCRIBE EXTENDED sales.orders;
Methods:
    - DESCRIBE DETAIL → Structured metadata (includes location, format, type).
    - DESCRIBE EXTENDED → Detailed metadata (includes Location field).
    - DESCRIBE TABLE → Only schema (no storage info).`

- Query Federation executes queries on external systems, while Catalog Federation runs queries on Databricks using external metadata.
Example:
    - Query Federation → SQL pushed to external DB (e.g., PostgreSQL executes query)
    - Catalog Federation → Databricks runs query using external catalog metadata
Methods:
    - Query Federation → Pushdown execution (external compute).
    - Catalog Federation → Databricks execution + external metadata access.
- Use Delta Sharing Open Sharing protocol to share data with external tools like Tableau without Databricks.
`Example: Share Delta table → partner connects via Tableau using shared credentials → queries data directly.
Methods:
    - Open Sharing → External tools (Tableau, Power BI, pandas) without Databricks.
    - Databricks-to-Databricks → Requires both sides to have Databricks.
    - Lakehouse Federation → For querying external data, not sharing.`
- Delta Sharing is read-only, meaning recipients cannot modify or write back to shared data.
Example: Partner queries shared table → can analyze/export data but cannot update/delete rows.
Methods:
    - Delta Sharing → Secure, live data access.
    - Read-only access → No INSERT/UPDATE/DELETE allowed.
    - Open Sharing vs Databricks-to-Databricks → Different capabilities, but both are read-only for recipients.
- Grant EXECUTE on the function to allow usage without modification.
`Example: GRANT EXECUTE ON FUNCTION projects.ml_utils.predict_score TO model_training;
Methods:
    - EXECUTE → Allows running the function (required).
    - USAGE → Needed on catalog/schema (not sufficient alone).
    - ALL PRIVILEGES → Too broad (avoid).`
- Cross-cloud or cross-region Delta Sharing can incur cloud egress charges for data transfer.`
Example: AWS (US-East) → Azure (EU-West) → data transfer billed as egress from AWS.
Methods:
    - Cross-region/cloud sharing → Triggers egress costs.
    - Same region sharing → No/low egress cost.
    - Compression → Reduces size, not cost elimination.`
        - Egress cost is the fee you pay when data leaves your cloud provider or region.
                - AWS (US-East) → AWS (US-East) → usually free
                - AWS (US-East) → AWS (EU-West) → charged
                - AWS → Azure → charged (cross-cloud)
