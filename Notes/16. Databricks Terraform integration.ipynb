# 📘 Databricks Terraform Integration — Complete Notes (Beginner → Advanced)

---

## 🔷 1. What is Terraform?

👉 Terraform is an **Infrastructure as Code (IaC) tool** used to **provision and manage infrastructure using code**

---

### 🧠 Simple Understanding

| Without Terraform | With Terraform |
|------------------|---------------|
| Manual setup ❌ | Automated setup ✅ |
| Error-prone ❌ | Consistent ✅ |
| Hard to scale ❌ | Easily scalable ✅ |

---

## 🔷 2. What Terraform is Used For

👉 Automate creation of:
- Clusters
- Jobs
- Notebooks
- Unity Catalog
- Permissions
- Workspaces

---

👉 Key Benefits:
- Reproducibility
- Version control (Git)
- CI/CD integration
- Environment management

---

## 🔷 3. Terraform Core Concepts

---

| Concept | Meaning |
|--------|--------|
| Provider | Plugin (Databricks) |
| Resource | Object (job, cluster) |
| State | Current infra status |
| Plan | Preview changes |
| Apply | Execute changes |

---

## 🔷 4. Databricks Terraform Provider

👉 Official provider to manage Databricks resources

---

### Example Setup

```hcl
provider "databricks" {
  host  = "https://adb-xxxx.azuredatabricks.net"
  token = "your-token"
}
```

---

## 🔷 5. Common Resources in Databricks Terraform

---

| Resource | Purpose |
|---------|--------|
| databricks_cluster | Create cluster |
| databricks_job | Create job |
| databricks_notebook | Upload notebook |
| databricks_sql_endpoint | SQL warehouse |
| databricks_permissions | Manage access |

---

## 🔷 6. Example: Create Cluster

```hcl
resource "databricks_cluster" "cluster" {
  cluster_name  = "demo-cluster"
  spark_version = "13.3.x-scala2.12"
  node_type_id  = "Standard_DS3_v2"
  num_workers   = 2
}
```

---

## 🔷 7. Example: Create Job

```hcl
resource "databricks_job" "job" {
  name = "etl-job"

  task {
    task_key = "task1"

    notebook_task {
      notebook_path = "/Workspace/etl"
    }

    existing_cluster_id = databricks_cluster.cluster.id
  }
}
```

---

## 🔷 8. Terraform Workflow

---

```
Write Code → terraform init → terraform plan → terraform apply
```

---

### Commands

```bash
terraform init
terraform plan
terraform apply
```

## 🔷 9. Authentication Methods

---

- Azure Service Principal
- AWS IAM Role

## 🔷 11. Terraform vs CLI vs Asset Bundles

---

| Feature | Terraform | CLI | Asset Bundles |
|--------|----------|-----|---------------|
| Infra setup | ✅ | ❌ | ❌ |
| Deployment | ✅ | ⚠️ | ✅ |
| CI/CD | ✅ | ⚠️ | ✅ |
| Use case | Infra | Commands | Pipelines |

## 🔷 13. Best Practices 🚀

---

- Store code in Git
- Use separate environments (dev/prod)
- Use variables instead of hardcoding
- Secure tokens

---

## 🔷 14. Common Mistakes 🚨

---

- ❌ Hardcoding credentials
- ❌ Not using state properly
- ❌ Mixing environments
- ❌ Using PAT for service principals
