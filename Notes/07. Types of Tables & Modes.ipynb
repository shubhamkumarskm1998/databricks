# 📘 Databricks Tables & Modes — Complete Guide

---

## 🔷 1. Types of Tables in Databricks

Databricks mainly supports **2 types of tables**:

---

## 1️⃣ Managed Tables

👉 Databricks **fully manages both data + metadata**

### 🔹 Key Features:
- Data stored in **default storage location**
- When table is dropped → **data is deleted**
- Easy to manage

### 🔹 Example (SQL)
```sql
CREATE TABLE sales.analytics.customers (
  id INT,
  name STRING
);
```

### 🔹 Example (PySpark)
```python
df.write.saveAsTable("sales.analytics.customers")
```

### 🔹 When to Use:
- Quick development
- No need to manage storage manually

---

## 2️⃣ External Tables

👉 Only metadata is managed by Databricks
👉 Data is stored **externally (S3 / ADLS / GCS)**

### 🔹 Key Features:
- Data remains even if table is dropped
- Full control over storage
- Useful for shared data lakes

### 🔹 Example (SQL)
```sql
CREATE TABLE sales.analytics.customers_ext
USING DELTA
LOCATION 's3://company-data/customers/';
```

### 🔹 Example (PySpark)
```python
df.write \
  .format("delta") \
  .option("path", "s3://company-data/customers/") \
  .saveAsTable("sales.analytics.customers_ext")
```

### 🔹 When to Use:
- Enterprise data lakes
- Shared storage systems

---

## 🔷 2. Temporary Tables / Views

### 🔹 Temporary View

👉 Exists only for **current session**

```sql
CREATE OR REPLACE TEMP VIEW temp_customers AS
SELECT * FROM sales.analytics.customers;
```

```python
df.createOrReplaceTempView("temp_customers")
```

---

### 🔹 Global Temporary View

👉 Accessible across notebooks (same cluster)

```sql
CREATE OR REPLACE GLOBAL TEMP VIEW global_customers AS
SELECT * FROM sales.analytics.customers;
```

👉 Access using:
```sql
SELECT * FROM global_temp.global_customers;
```

---

## 🔷 3. Delta Tables (Most Important 🔥)

👉 Default format in Databricks

### 🔹 Features:
- ACID transactions
- Schema enforcement
- Time travel
- Versioning

### 🔹 Example
```sql
CREATE TABLE sales.analytics.customers_delta
USING DELTA;
```

```python
df.write.format("delta").saveAsTable("sales.analytics.customers_delta")
```

---

## 🔷 4. Write Modes in Databricks

### 🔹 Append Mode
```python
df.write.mode("append").saveAsTable("table_name")
```

### 🔹 Overwrite Mode
```python
df.write.mode("overwrite").saveAsTable("table_name")
```

### 🔹 Error Mode (Default)
```python
df.write.mode("error").saveAsTable("table_name")
```

### 🔹 Ignore Mode
```python
df.write.mode("ignore").saveAsTable("table_name")
```

---

## 🔷 5. SQL Equivalent Modes

### Insert (Append)
```sql
INSERT INTO table_name SELECT * FROM source_table;
```

### Overwrite
```sql
INSERT OVERWRITE table_name SELECT * FROM source_table;
```

---

## 🔷 6. Key Differences

| Feature | Managed Table | External Table |
|--------|--------------|---------------|
| Storage | Databricks managed | External |
| Drop Table | Deletes data | Keeps data |
| Control | Less | Full |
| Use Case | Dev/Test | Production |

---

## 🔷 7. Interview-Level Points

- Managed → Databricks controls data
- External → User controls data
- Default format → Delta
- Append → Adds data
- Overwrite → Replaces data

---

## 🔷 🚀 One-Line Summary

> Databricks tables = Managed + External + Delta with flexible write modes
