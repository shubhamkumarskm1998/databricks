### Delta Lake

### Who Created Delta Lake?
- Created by Databricks
- Open-sourced in 2019
- Built on top of Apache Spark

---

### What is Delta Lake?
- Storage layer on top of Data Lake
- Adds reliability and performance to data lakes
- Uses Parquet + transaction log

---

### Problems in Data Lakes (Before Delta Lake)

- No ACID transactions → data corruption possible
- No schema enforcement → inconsistent data
- No version control → cannot rollback
- Poor performance for updates/deletes
- Batch vs Streaming separation

---

### What Delta Lake Solved

#### 1. ACID Transactions
- Ensures reliable reads/writes
- Prevents data corruption

#### 2. Schema Enforcement
- Maintains data consistency
- Prevents bad data ingestion

#### 3. Time Travel (Versioning)
- Access previous versions of data
- Enables rollback and auditing

#### 4. Unified Batch + Streaming
- Same table supports both workloads

#### 5. Efficient Updates / Deletes
- Supports MERGE, UPDATE, DELETE operations

#### 6. Performance Improvements
- File compaction (OPTIMIZE)
- Data skipping
- Indexing improvements

---

### Key Components

- Delta Table → stored as Parquet files
- Transaction Log (_delta_log) → tracks all changes
- Metadata → schema, versions, operations

---

### Key Insight
Delta Lake = Data Lake + Reliability + Performance

---

### One-line Summary
Delta Lake, created by Databricks, solves data lake reliability issues using ACID transactions, schema enforcement, and versioning

### 1. Data Lake
- Stores raw data (structured + unstructured)
- Cheap storage (S3, ADLS, GCS)
- No ACID transactions
- No schema enforcement
- Example → CSV, JSON, Parquet files

👉 Problem → Dirty data, slow queries, no reliability

---

### 2. Delta Lake
- Storage layer on top of Data Lake

  - ACID transactions

👉 Supports:

---

### Key Insight
Data Lake → storage

---

### One-line Summary
Lakehouse = Data Lake + Delta Lake + Compute + Governance

# Data Warehouse vs Data Lake vs Data Lakehouse

---

## 📊 What is Data Warehouse?

### Architecture
Operational Data + External Data → ETL → Data Warehouse → Data Marts → BI Reports

### Limitations
- Increased data volume & variety
- Longer time to ingest new data
- Proprietary formats (vendor lock-in)
- Scalability issues
- High storage costs

---

## 🌊 What is Data Lake?

### Architecture
Operational Data + External Data → ELT → Data Lake → ELT → Data Warehouse → BI Reports
Also supports AI/ML workloads directly

### Limitations
- No ACID transactions
- Partial data loads
- Inconsistent reads
- Difficult data corrections
- No rollback capability
- GDPR challenges
- No data versioning
- Poor BI performance
- Complex setup
- Streaming vs Batch complexity

---

## 🏗️ What is Data Lakehouse?

### Architecture
Operational Data + External Data → ELT → Data Lakehouse → BI Reports
Also supports AI/ML workloads

### Advantages
- Handles all types of data
- Cheap cloud object storage
- Uses open file formats
- Supports all workloads (BI + ML + AI)
- Direct BI integration
- ACID transactions & version history
- Improved performance
- Simple architecture

---

## 🔥 Key Comparison

| Feature | Data Warehouse | Data Lake | Lakehouse |
|--------|--------------|----------|----------|
| Data Type | Structured | All | All |
| Cost | High | Low | Medium |
| Performance | High | Low | High |
| ACID | Yes | No | Yes |
| Schema | Strict | None | Enforced |
| ML Support | Limited | Yes | Yes |
| BI Support | Yes | Weak | Strong |

---

## 🎯 One-line Summary

- Data Warehouse → Structured analytics
- Data Lake → Raw data storage
- Lakehouse → Unified modern architecture

👉 Lakehouse = Best of both worlds

# 🏗️ Medallion Architecture (Databricks)

## 📌 What is Medallion Architecture?

A data design pattern that organizes data into layers to improve:
- Data quality
- Reliability
- Performance

👉 Layers:
Bronze → Silver → Gold

---

## 🥉 1. Bronze Layer (Raw Data)

### 📌 Purpose
- Store raw, unprocessed data
- Ingest data from multiple sources

### 📊 Characteristics
- No transformations
- Append-only
- Schema may be messy
- Keeps full history

### 📥 Sources
- APIs
- Logs
- CSV / JSON files
- Streaming data (Kafka)

### 💡 Example

Raw JSON:
{
  "id": 1,
  "name": "Alice",
  "amount": "100",
  "timestamp": "2024-01-01T10:00:00"
}

Stored as-is in Delta table

---

## 🥈 2. Silver Layer (Cleaned Data)

### 📌 Purpose
- Clean and transform data
- Apply schema
- Remove duplicates

### 📊 Transformations
- Data cleaning
- Type casting
- Deduplication
- Joins

### 💡 Example

Before:
amount = "100" (string)

After:
amount = 100 (integer)

SQL Example:
SELECT
  id,
  name,
  CAST(amount AS INT) as amount,
  timestamp
FROM bronze_table
WHERE id IS NOT NULL

---

## 🥇 3. Gold Layer (Business Data)

### 📌 Purpose
- Aggregated, business-level data
- Ready for BI / dashboards

### 📊 Transformations
- Aggregations
- KPIs
- Business logic

### 💡 Example

Total sales per user:
SELECT
  name,
  SUM(amount) as total_sales
FROM silver_table
GROUP BY name

---

## 🔁 Data Flow

Bronze → Silver → Gold

Each layer improves:
- Quality ↑
- Structure ↑
- Business value ↑

---

## ⚙️ Real Example (E-commerce)

### Bronze
- Raw orders data (JSON logs)

### Silver
- Cleaned orders
- Valid users only
- Correct data types

### Gold
- Revenue per day
- Top customers
- Sales dashboard

---

## 🔥 Benefits

- Data quality improvement
- Easy debugging (trace back to raw)
- Scalability
- Reusability of data
- Separation of concerns

---

## ⚠️ Best Practices

- Use Delta tables
- Enable schema enforcement
- Use partitioning
- Maintain lineage (Unity Catalog)

---

## 🎯 One-line Summary

Medallion Architecture = Layered data pipeline (Bronze → Silver → Gold) for clean, reliable analytics

- Databricks UI (workspace)

- Job orchestration
- Unity Catalog (governance)
- Query & code management

### 🔐 Key Point:
- Runs in Databricks account
- You don’t manage it

---

## 🟩 2. Compute Plane (Customer Managed)

### 📌 Types:

#### a) Classic Compute
- Your clusters (VMs)
- Runs Spark jobs
- Data processing happens here

#### b) Serverless Compute
- Managed by Databricks
- No cluster setup
- Auto-scaling

### Workspace Cloud Storage
- Stores:
  - DBFS
  - Tables (Delta Lake)
  - Files

---

## 🧱 4. Customer Resources

- Databases
- Data lakes (S3 / ADLS / GCS)
- External systems

---

## 🔄 Flow

User → Control Plane → Compute Plane → Storage

---

1. User runs notebook

3. Control Plane sends job to Compute Plane

- Control plane never stores your data

- Unity Catalog controls access

---

# ⚙️ Databricks Classic Compute

## 🧠 What is Classic Compute?
- Compute runs in customer cloud
- Uses clusters (VMs)
- Executes Spark jobs

---

## 🔀 Types of Clusters

### 1. All-Purpose Cluster

#### 📌 Characteristics
- Created manually
- Persistent (runs until stopped)
- Used for interactive workloads
- Shared among users
- Expensive

#### 💡 Use Cases
- Notebooks
- Data exploration
- Development

---

### 2. Job Cluster

#### 📌 Characteristics
- Created automatically by jobs
- Temporary (auto-terminated)
- Used for automated workloads
- Isolated per job
- Cheaper

#### 💡 Use Cases
- Scheduled pipelines
- ETL jobs
- Batch processing

---

## 🔥 Key Differences

| Feature | All-Purpose | Job Cluster |
|--------|------------|------------|
| Creation | Manual | Automatic |
| Lifetime | Persistent | Temporary |
| Usage | Interactive | Automated |
| Sharing | Shared | Isolated |
| Cost | High | Lower |

---

## 🔄 Example

### All-Purpose
- Analyst runs notebook daily
- Cluster stays ON → expensive

### Job Cluster
- ETL runs at 2 AM
- Cluster starts → runs → stops → cost saved

---

## 🎯 One-line Summary

All-Purpose = Interactive + Persistent
Job Cluster = Automated + Temporary

---

- Defines VM size
- Example:

- Master node
- Coordinates execution

- Automatically adjusts number of workers

- Stops cluster after inactivity

# 💻 %sh runs only on Driver Node (Meaning)

---

## 🧠 What is %sh?

- %sh is used to run **Linux shell commands**
- Example:
  %sh
  ls -l

---

## 🧱 Cluster Structure

A Databricks cluster has:

- Driver Node → Controls execution (brain 🧠)
- Worker Nodes → Execute distributed tasks (muscles 💪)

---

## 📌 What does "runs only on driver node" mean?

👉 When you use %sh:
- Command runs ONLY on the **driver machine**
- It does NOT run on worker nodes

---

## 💡 Example

%sh
hostname

👉 Output will show:
- Driver machine name only
- Not worker machines

---

## ⚠️ Why not workers?

- %sh is meant for:
  - Debugging
  - File inspection
  - Installing system-level tools

👉 Workers are managed separately by Spark

---

## ❌ Common Mistake

Thinking:
%sh will run on all nodes ❌

👉 It runs only on driver ✅

---

## 🔥 Real Problem Example

%sh
pip install some_package

👉 Installed only on driver ❌
👉 Workers will NOT have it → Spark job may fail

---

## ✅ Correct Approach

Use:
%pip install package_name

👉 Installs on:
- Driver
- Workers

---

## 🎯 Use Cases of %sh

- Check files:
  %sh ls /dbfs/

- Debug environment:
  %sh env

- Run shell scripts

---

## 🚫 When NOT to use %sh

- Installing libraries for Spark jobs
- Running distributed commands

---

## 🎯 One-line Summary

%sh runs shell commands only on the driver node, not on worker nodes

# 🛠️ Databricks Utilities (dbutils)

---

## 🧠 What is dbutils?

- Built-in utility in Databricks
- Helps interact with:
  - File system
  - Secrets
  - Notebooks
  - Widgets

---

## 📂 1. File System Utilities (dbutils.fs)

### 📌 Purpose
- Work with DBFS (Databricks File System)

### 🔧 Commands

- List files:
  dbutils.fs.ls("/mnt/data")

- Copy files:
  dbutils.fs.cp("source_path", "destination_path")

- Move files:
  dbutils.fs.mv("source", "destination")

- Remove files:
  dbutils.fs.rm("path", recurse=True)

---

## 🔐 2. Secrets Utilities (dbutils.secrets)

### 📌 Purpose
- Securely store credentials
- Avoid hardcoding passwords

### 🔧 Example

dbutils.secrets.get(scope="my_scope", key="db_password")

---

## 🎛️ 3. Widgets Utilities (dbutils.widgets)

### 📌 Purpose
- Create input parameters in notebooks

### 🔧 Example

- Create widget:
  dbutils.widgets.text("name", "default_value")

- Get value:
  dbutils.widgets.get("name")

---

## 🔄 4. Notebook Utilities (dbutils.notebook)

### 📌 Purpose
- Control notebook workflows

### 🔧 Example

- Run another notebook:
  dbutils.notebook.run("/Shared/etl", 60)

- Exit notebook:
  dbutils.notebook.exit("Success")

---

## 🔥 Real Use Cases

- Read/write data from DBFS
- Secure API keys using secrets
- Pass parameters between jobs
- Orchestrate pipelines

---

## ⚠️ Best Practices

- Always use secrets for credentials
- Avoid hardcoding paths
- Use widgets for dynamic pipelines

---

## ❌ Common Mistakes

- Storing passwords in code ❌
- Not using widgets for parameters ❌
- Misusing file paths (/dbfs vs dbfs:/)

---

## 🎯 One-line Summary

dbutils provides utilities for file handling, secrets management, parameterization, and workflow control in Databricks

---

### 2. Step Execution

- Step Over → Next line
- Step Into → Enter function
- Step Out → Exit function

---

### 3. Variable Explorer

- View variable values in real-time
- Track changes during execution

---

### 4. Debug Console

- Run commands while paused
- Inspect state of program

---

## 💡 Example

def add(a, b):
    return a + b

result = add(5, 3)

👉 Add breakpoint inside function
👉 Step through execution
👉 Check values of a and b

---

## 🔥 Use Cases

- Debug ETL pipelines
- Fix transformation errors
- Understand complex logic
- Inspect data issues

---

## ⚠️ Limitations

- Works only with Python
- Requires supported runtime
- Not available for all cluster types

---

## ❌ Common Mistakes

- Not enabling debugger ❌
- Trying to debug SQL directly ❌
- Ignoring breakpoints ❌

---

## 🎯 One-line Summary

Databricks debugging allows step-by-step execution of Python code using breakpoints, variable inspection, and a debug console
