# 🔷 5. IF / ELSE Conditions

---

## 🔹 Concept

👉 Run tasks based on condition

---

## 🔹 Example Scenario

👉 If data exists → run processing
👉 Else → skip / run fallback

---

## 🔹 PySpark Example (Inside Notebook)

```python
df = spark.read.table("sales")

if df.count() > 0:
    print("Data available → process")
else:
    print("No data → skip")
```

---

## 🔹 SQL Example

```sql
SELECT CASE
    WHEN COUNT(*) > 0 THEN 'PROCESS'
    ELSE 'SKIP'
END AS decision
FROM sales;
```

---

## 🔹 Job-Level Condition (Concept)

👉 In Databricks UI:
- Task B runs only if Task A succeeds or meets condition

# 🔷 6. FOR EACH (Loop / Parallel Processing)

---

## 🔹 Concept

👉 Run same task for multiple inputs

---

## 🔹 Example Scenario

👉 Process multiple tables

---

## 🔹 PySpark Example

```python
tables = ["customers", "orders", "products"]

for table in tables:
    df = spark.read.table(table)
    df.write.mode("overwrite").saveAsTable(f"{table}_processed")
```

# 🔷 7. foreachBatch (Streaming Jobs 🔥)

---

## 🔹 Concept

👉 Apply batch logic on streaming data

---

## 🔹 PySpark

```python
def process_batch(df, batch_id):
    df.write.mode("append").saveAsTable("output")

df.writeStream.foreachBatch(process_batch).start()
```

# 🔷 15. Advanced Scenario (Loop + Merge)

---

```python
dates = ["2024-01-01", "2024-01-02"]

from delta.tables import DeltaTable
delta = DeltaTable.forName(spark, "target")

for d in dates:
    df = spark.read.table("source").filter(f"date = '{d}'")

    delta.alias("t").merge(
        df.alias("s"),
        "t.id = s.id"
    ).whenMatchedUpdateAll() \
     .whenNotMatchedInsertAll() \
     .execute()
```
