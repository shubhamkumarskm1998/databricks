# 📘 Databricks CLI & Asset Bundles — Complete Detailed Notes (Beginner → Advanced)

---

## 🔷 1. What is Databricks CLI?

👉 Databricks CLI is a **command-line tool** to interact with Databricks from terminal

👉 Used for:
- Managing jobs
- Uploading notebooks
- Managing files (DBFS)
- Automating workflows

---

### 🧠 Simple Understanding

Without CLI:
- Manual UI operations ❌

With CLI:
- Automation via scripts ✅
- CI/CD integration ✅

---

## 🔷 2. Installation

---

```bash
pip install databricks-cli
```

---

## 🔷 3. Configure CLI

---

```bash
databricks configure --token
```

---

👉 Requires:
- Workspace URL
- Personal Access Token

---

## 🔷 4. Common CLI Commands

---

## 🔹 1. Workspace Commands

```bash
databricks workspace list /
databricks workspace import notebook.py /Workspace/notebook.py
databricks workspace export /Workspace/notebook.py notebook.py
```

---

## 🔹 2. DBFS Commands

```bash
databricks fs ls dbfs:/
databricks fs cp file.txt dbfs:/data/file.txt
```

---

## 🔹 3. Jobs Commands

```bash
databricks jobs list
databricks jobs run-now --job-id 123
```

---

## 🔹 4. Clusters Commands

```bash
databricks clusters list
databricks clusters start --cluster-id 123
```

---

## 🔹 5. Secrets Commands

```bash
databricks secrets list-scopes
databricks secrets put --scope my-scope --key my-key
```

---

## 🔷 5. Use Cases of CLI

---

- Automate deployments
- Upload notebooks
- Trigger jobs
- Integrate with CI/CD (GitHub, Azure DevOps)

# 📦 Databricks Asset Bundles (VERY IMPORTANT 🔥)

---

## 🔷 6. What are Asset Bundles?

👉 Asset Bundles are used to **package and deploy Databricks resources**

👉 Includes:
- Jobs
- Notebooks
- Pipelines
- Configurations

---

### 🧠 Simple Understanding

Without Bundles:
- Manual deployment ❌

With Bundles:
- Automated deployment ✅
- Version-controlled pipelines ✅

---

## 🔷 7. Why Use Asset Bundles?

---

- CI/CD support
- Environment management
- Reproducible deployments

---

## 🔷 8. Bundle Structure

---

```
bundle/
  ├── databricks.yml
  ├── resources/
  │     ├── job.yml
  │     ├── pipeline.yml
  ├── src/
  │     ├── notebook.py
```

---

## 🔷 9. databricks.yml (Core File)

---

👉 Defines:
- Workspace
- Jobs
- Pipelines

---

### Example

```yaml
bundle:
  name: my_bundle

targets:
  dev:
    workspace:
      host: https://adb-123.4.azuredatabricks.net

resources:
  jobs:
    my_job:
      name: my-job
      tasks:
        - task_key: task1
          notebook_task:
            notebook_path: ./src/notebook.py
```

---

## 🔷 10. Deployment Commands

---

```bash
databricks bundle validate
databricks bundle deploy
databricks bundle run my_job
```

---

## 🔷 11. Environment Management

---

👉 Multiple environments:

```yaml
targets:
  dev:
  prod:
```

---

👉 Allows:
- Separate configs
- Safe deployments

---

## 🔷 12. Real-World Workflow

---

1. Write notebook/code
2. Define job in YAML
3. Validate bundle
4. Deploy to workspace
5. Run job

---

## 🔷 13. Example: Job Deployment

---

### YAML

```yaml
resources:
  jobs:
    etl_job:
      name: etl-job
      tasks:
        - task_key: load_data
          notebook_task:
            notebook_path: ./src/load.py
```

## 🔷 15. CLI vs Asset Bundles

---

| Feature | CLI | Asset Bundles |
|--------|-----|---------------|
| Use | Commands | Deployment |
| Automation | Medium | High |
| CI/CD | Supported | Native |
| Structure | Script-based | YAML-based |
