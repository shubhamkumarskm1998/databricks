- Apache Spark

## 🔹 Databricks Offerings

Databricks provides 2 main offerings:

## 1️⃣ Databricks Platform

👉 Cloud-based Spark platform with easy UI

### Key Features:

- Launch **managed Spark clusters**
- Use **notebooks** (Python, SQL, etc.)
- Create **workspaces** for team collaboration
- Role-Based Access Control (**RBAC**)

  - On demand
  - Scheduled

👉 Free version for learning

## 📌 What is Databricks?

**Databricks** is a **cloud-based data platform** built on top of **Apache Spark** that provides a unified environment for:

- Data Engineering
- Data Science
- Machine Learning
- Data Analytics

👉 It is known as a **Lakehouse Platform** (combines Data Lake + Data Warehouse).

---

## 🤔 Why was Databricks created?

Before Databricks, organizations faced multiple problems:

### ❌ Traditional Challenges

- Separate tools for:
  - Data processing (Hadoop)
  - Data warehousing (Snowflake, Redshift)
  - Machine learning
- Complex cluster setup & maintenance
- Slow data pipelines
- Poor collaboration between teams
- Difficulty handling **big data at scale**

---

### 💡 Solution by Databricks

Databricks was created to:

- Simplify **big data processing**
- Provide **one unified platform**
- Remove infrastructure complexity
- Enable **real-time + batch processing**
- Improve collaboration across teams

---

# ☁️ Introduction to Azure Databricks

## 🔹 What is Azure Databricks?

👉 Azure Databricks is a **partnership between Microsoft and Databricks**

👉 It provides:
- Databricks Platform inside **Azure Portal**
- Runs on **Azure compute services**

---

## 🔹 Environments in Azure Databricks

Azure Databricks provides **2 main environments**:

---

## 1️⃣ Azure Databricks Workspace

👉 Main working environment

### Key Points:

- Integrated with **Azure data services**
- Used for building **data pipelines**

---

### 🔗 Integrations:

#### Batch Processing
- Azure Data Factory

#### Streaming
- Apache Kafka
- Azure Event Hub
- IoT Hub

#### Storage
- Azure Blob Storage
- Azure Data Lake Storage

---

## 2️⃣ Azure Databricks SQL Analytics

👉 Used for **querying and reporting**

### Key Points:

- Run **SQL queries** on data lake
- Create **visualizations (charts, dashboards)**
