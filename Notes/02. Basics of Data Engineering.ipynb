# 🚀 Basics of Data Engineering

---

## 📌 What is Data Engineering?

**Data Engineering** is the process of designing, building, and maintaining systems that:

- Collect data
- Store data
- Process data
- Make data available for analysis

👉 Goal: Deliver **clean, reliable, and usable data** for analytics & ML

---

## 🔄 Data Engineering Lifecycle
Data Sources → Ingestion → Storage → Processing → Serving → Analytics


### Steps Explained:

1. **Data Sources**
   - Databases (MySQL, PostgreSQL)
   - APIs
   - Logs
   - IoT devices

2. **Ingestion**
   - Batch (scheduled loads)
   - Streaming (real-time)

3. **Storage**
   - Data Lakes (raw data)
   - Data Warehouses (structured data)

4. **Processing**
   - Cleaning
   - Transformations
   - Aggregations

5. **Serving**
   - Data exposed for:
     - BI tools
     - ML models

6. **Analytics**
   - Dashboards
   - Reports
   - Predictions

---

## 🏗️ Types of Data Processing

### 1. Batch Processing
- Processes large chunks of data at intervals
- Example: Daily ETL job

### 2. Stream Processing
- Processes data in real-time
- Example: Fraud detection

```sql
SELECT * FROM employees;

### Query Languages:
- SQL
- GraphQL
- Kusto

### Batch vs Stream Processing
- Batch Processing
    - Data processed in chunks
    - Scheduled (daily, hourly)
    - Cost-effective
    - Not real-time

- Stream Processing
    - Real-time processing
    - Continuous data flow
    - Producers → send data
    - Consumers → read data
    - More expensive

### Pivot Table
- Summarizes large data into statistics
- Used for:
- Aggregations (sum, avg)
- Grouping data
- Quick insights

### Data Concepts (Important for Exams)
    - Data Modelling -> Designing structure of data
    - Data Integrity -> Ensuring data is correct & consistent
    - Normalization -> Reduce redundancy (more accuracy)
    - Denormalization -> Improve performance (faster queries)

## 📊 8. Pivot

👉 Used to **summarize data**

Example:
Delhi → 100, 200
👉 Pivot → Delhi = 300

✔ Groups + calculates (SUM, COUNT, AVG)

---

## 🗄️ 9. Table

👉 Data stored in rows & columns

Example:

| ID | Name    |
|----|---------|
| 1  | Shubham |

---

## 🔑 10. Keys

### Primary Key
👉 Unique ID
Example: ID = 1,2,3

### Foreign Key
👉 Connect tables
Example: Order → CustomerID

---

## 🔗 11. Relationships

- One to One → person → passport
- One to Many → customer → orders
- Many to Many → student ↔ course

---

## ⚡ 12. Index

👉 Makes search fast

Example:
- Without index → slow
- With index → fast

👉 Like book index

---

## 👁️ 13. View

👉 Saved query (virtual table)

Does NOT store data

---

## 🧱 14. Row vs Column

### Row Store
👉 Stored row-wise
✔ Good for apps

### Column Store
👉 Stored column-wise
✔ Good for analytics

---

## 🧠 15. Important Concepts

- Data Modeling → design tables
- Data Integrity → correct data
- Normalization → remove duplicates
- Denormalization → faster queries

---

## 🎯 Final Summary

👉 Data Engineering =
Collect → Store → Process → Analyze data

# 🚀 Data Engineering Notes

---

# 📦 Non-Relational Data (NoSQL)

👉 A non-relational database stores data in **non-tabular form**
👉 It is flexible and designed for different data structures

---

## 🔹 Types of Non-Relational Databases

### 1. Key-Value
- Data stored as **key → value pairs**
- Very fast and simple lookups
- Scales easily

Example:
user_id → Shubham

---

### 2. Document
- Data stored as **JSON-like documents**
- Flexible structure

Example:
{
  "name": "Shubham",
  "age": 25
}

---

### 3. Columnar
- Data stored **column-wise instead of row-wise**
- Good for large-scale analytics

---

### 4. Graph
- Data stored as **nodes and relationships**
- Used where relationships matter

Example:
User → Friend → User

---

## 🔑 Key Point

👉 Non-relational = flexible schema (no fixed table structure)

---

# 🛡️ Data Integrity

👉 Data integrity = **data is correct and consistent**

### Definition:
- Ensures data is stored exactly as intended
- Maintains accuracy throughout lifecycle

---

## 🔹 Why it is important?

- Prevent wrong data
- Ensure trust in data
- Avoid bad decisions

---

## 🔹 How to Ensure Data Integrity?

- Proper data modeling
- Use constraints (rules)
- Maintain data versions (backup)
- Validate data before storing
- Limit access (security)
- Use hash/check mechanisms

---

## ❌ Data Corruption

👉 Data corruption = **data becomes incorrect or damaged**

---

### 🔹 Causes of Data Corruption

- Hardware failure
- Human error
- Bugs in code
- Malicious attacks
- Unexpected system behavior

---

## 🔹 Impact

- Data loss
- Wrong insights
- System failure

---

## 🧠 Final Understanding

👉 Non-relational DB → flexible data storage
👉 Data Integrity → data should be correct
👉 Data Corruption → data becomes wrong

---

# 🧱 Normalized vs Denormalized Data

## 🔹 Normalized Data

👉 Data is stored in a **structured and separate way**
👉 Focus: **No duplication + consistency**

### Key Points:
- Data is **non-redundant** (no duplicates)
- Data is **consistent**
- Uses **multiple tables**
- Saves storage space
- Maintains **data integrity**

### Example:

Products Table
| id | name | category_id |

Categories Table
| id | name |

Orders Table
| id | product_id | customer_id |

Customers Table
| id | name |

👉 Data is split into different tables and linked

---

## 🔹 Denormalized Data

👉 Data is **combined into one table**
👉 Focus: **Fast querying**

### Key Points:
- Data is **duplicated (redundant)**
- Fewer tables
- Faster reads (queries)
- More storage used
- Data integrity is weaker

### Example:

Customer Orders Table
| product_name | category_name | customer_name | order_amount |

👉 Everything stored in one table

---

## 🔥 Key Difference

| Normalized | Denormalized |
|----------|-------------|
| No duplicates | Duplicate data |
| More tables | Fewer tables |
| Slower queries | Faster queries |
| High integrity | Lower integrity |

---

# ⚖️ Data Consistency

## 🔹 What is Data Consistency?

👉 When same data exists in multiple places,
👉 it should **match exactly**

---

## 🔹 Strongly Consistent

👉 Always returns **latest correct data**

### Key Points:
- Data is always up-to-date
- No stale (old) data
- May be slightly slower

### Example:
- Bank balance check → always correct

---

## 🔹 Eventually Consistent

👉 Data may be **temporarily different**, but becomes correct later

### Key Points:
- Fast response
- May return old data initially
- Becomes correct after some time

### Example:
- Social media likes count
- You may see 100 → later becomes 120

---

## 🔥 Key Difference

| Strong Consistency | Eventual Consistency |
|------------------|---------------------|
| Always correct | Eventually correct |
| Slower | Faster |
| No stale data | Temporary stale data |

---

## 🧠 Final Understanding

👉 Normalization → accuracy
👉 Denormalization → speed

👉 Strong consistency → correct now
👉 Eventual consistency → correct later

---

## 🔄 What is Synchronous vs Asynchronous?

👉 Refers to how **data is transmitted or replicated**

---

## 🔹 Synchronous

👉 Data transfer happens **in real-time (in sync)**

### Key Points:
- Uses **clock/timer** (time is guaranteed)
- Data is sent as a **continuous stream**
- You get data **only after transfer is complete**
- Always **consistent (latest data)**
- **Slower**

---

### Example:

👉 Bank system
- When you transfer money → balance updates immediately
- No data loss allowed

---

### Use Case:
- Backup systems
- Critical systems (finance, banking)

---

## 🔹 Asynchronous

👉 Data transfer happens **with delay**

### Key Points:
- No strict timing guarantee
- Data sent in **parts (start/stop)**
- You can access data anytime
- May get **old/stale data**
- **Faster**

---

### Example:

👉 Social media
- Likes may show 100 → later update to 120

---

### Use Case:
- Analytics systems
- Read replicas
- Reporting systems

---

## 🔥 Key Difference

| Synchronous | Asynchronous |
|------------|-------------|
| Real-time | Delayed |
| Always consistent | Eventually consistent |
| Slower | Faster |
| No data loss | Possible temporary mismatch |

---

## 🧠 Final Understanding

👉 Synchronous = **accuracy first**
👉 Asynchronous = **speed first**

---

# 📡 Data Sources

## 🔹 What is a Data Source?

👉 A data source is **where data originates from**

---

## 🔹 Examples of Data Sources

- Data Lake
- Data Warehouse
- Database
- Datastore
- API (web app data)
- Flat files (Excel, CSV)

---

## 🔹 Key Idea

👉 Tools (like BI tools) connect to data sources to:
- Extract data
- Create reports
- Build dashboards

---

# 🗄️ Datastore

## 🔹 What is Datastore?

👉 A datastore is a **place to store and manage data**

- Stores **unstructured or semi-structured data**
- Broad term (can include databases)

---

## 🔹 Examples

- Files (documents, spreadsheets)
- Email systems
- Databases (SQL / NoSQL)
- Distributed storage systems

---

# 🗃️ Database

## 🔹 What is a Database?

👉 A database is a **structured data store**

- Stores **structured or semi-structured data**
- Uses proper design (tables, schema)

---

## 🔹 Types

### Relational Database
- Data in **tables (rows & columns)**

### Non-Relational Database
- Flexible structure (JSON, key-value, etc.)

---

## 🔹 Features

- Query using SQL
- Data modeling
- Optimized data retrieval

---

# 🏢 Data Warehouse

## 🔹 What is Data Warehouse?

👉 A data warehouse is designed for **analytics (reporting)**

---

## 🔹 Key Points

- Stores **large amount of data**
- Optimized for **queries (fast reads)**
- Usually **column-based storage**
- Performs **aggregation (SUM, AVG)**

---

## 🔹 Usage

- Business reports
- Dashboards
- Historical analysis

---

## 🔹 Important

👉 Not used for real-time operations
👉 Used periodically (daily/weekly reports)

---

# 🧩 Data Mart

## 🔹 What is Data Mart?

👉 A data mart is a **subset of a data warehouse**

---

## 🔹 Key Points

- Smaller (usually < 100 GB)
- Focus on **one business area**
  - Sales
  - Marketing
- Often **read-only**
- Faster and cheaper queries

---

# 🌊 Data Lake

## 🔹 What is Data Lake?

👉 A data lake is a **storage for raw data**

---

## 🔹 Key Points

- Stores **huge amount of data**
- Supports:
  - Structured
  - Semi-structured
  - Unstructured data

- No strict schema required

---

## 🔹 Use Cases

- Machine Learning
- Big data analytics
- Real-time processing

---

## ⚠️ Problem

👉 If not managed → becomes **Data Swamp** (messy data)

---

# 🏗️ Data Lakehouse

## 🔹 What is Data Lakehouse?

👉 Combines **Data Lake + Data Warehouse**

---

## 🔹 Key Idea

👉 Get benefits of both:
- Flexibility (lake)
- Performance (warehouse)

---

## 🔹 Features

- Supports:
  - BI (reports)
  - Machine Learning
  - Streaming

- Works with many data formats
- Easier to manage than data lake

---

## 🔹 Example

👉 Apache Delta Lake

---

# 🧠 Final Understanding

👉 Data Source → where data comes from
👉 Datastore → where data is stored
👉 Database → structured storage
👉 Warehouse → analytics
👉 Data Mart → smaller warehouse
👉 Data Lake → raw data
👉 Lakehouse → best of both worlds

---

# 🧱 Data Structures

## 🔹 What is a Data Structure?

👉 A data structure is **how data is organized and stored**

- Uses a **specific format**
- Makes data:
  - Easy to access
  - Easy to modify

👉 It can store different types of data

---

## 🔹 Types of Data Structure (Based on Organization)

Data can be divided into 3 types:

---

## 1️⃣ Unstructured Data

👉 Data with **no proper format or structure**

### Key Points:
- Random / messy data
- Hard to search or analyze
- No fixed schema

### Examples:
- Images
- Videos
- PDFs
- Text files

---

## 2️⃣ Semi-Structured Data

👉 Data with **some structure but not strict**

### Key Points:
- Has tags or keys
- Can be searched (limited)
- Flexible format

### Examples:
- JSON
- XML

Example:
```xml
<Customer>
  <Order>
    <Name>Item</Name>
    <Quantity>100</Quantity>
  </Order>
</Customer>
