## 🔷 1. Language Magic Commands

| Command | Example |
|--------|--------|
| %sql | ```sql<br>%sql<br>SELECT * FROM sales.analytics.customers;<br>``` |
| %python | ```python<br>%python<br>df = spark.table("sales.analytics.customers")<br>display(df)<br>``` |
| %scala | ```scala<br>%scala<br>val df = spark.table("sales.analytics.customers")<br>display(df)<br>``` |
| %r | ```r<br>%r<br>df <- SparkR::sql("SELECT * FROM sales.analytics.customers")<br>display(df)<br>``` |

## 🔷 2. File System Commands (%fs)

| Operation | Example |
|----------|--------|
| List Files | ```bash<br>%fs ls /FileStore/<br>``` |
| Read File | ```bash<br>%fs head /FileStore/data.txt<br>``` |
| Create Directory | ```bash<br>%fs mkdirs /FileStore/new_folder<br>``` |
| Delete File | ```bash<br>%fs rm /FileStore/data.txt<br>``` |

## 🔷 4. dbutils.fs (File Utilities)

| Operation | Example |
|----------|--------|
| List | ```python<br>dbutils.fs.ls("/FileStore/")<br>``` |
| Copy | ```python<br>dbutils.fs.cp("/FileStore/file1.txt", "/FileStore/file2.txt")<br>``` |
| Move | ```python<br>dbutils.fs.mv("/FileStore/file1.txt", "/FileStore/archive/file1.txt")<br>``` |
| Delete | ```python<br>dbutils.fs.rm("/FileStore/file1.txt", True)<br>``` |
| Read | ```python<br>dbutils.fs.head("/FileStore/file1.txt")<br>``` |
