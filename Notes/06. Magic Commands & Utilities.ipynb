# ⚡ Databricks Magic Commands & Utilities

---

## 🧠 Section 1: Magic Commands Overview

- Magic commands are used to **run different languages inside notebooks**
- Enable **multi-language workflows** in a single notebook
- Useful for:
  - Quick execution
  - Debugging
  - Exploration

👉 Syntax:
```sql
%command
```

---

## ⚙️ Section 2: Language Magic Commands

- `%sql` → Run SQL queries
```sql
%sql
SELECT * FROM sales.analytics.customers;
```

- `%python` → Run Python code
```python
%python
df = spark.table("sales.analytics.customers")
display(df)
```

- `%scala` → Run Scala code
```scala
%scala
val df = spark.table("sales.analytics.customers")
display(df)
```

- `%r` → Run R code
```r
%r
df <- SparkR::sql("SELECT * FROM sales.analytics.customers")
display(df)
```

## 🔄 Section 3: File System & Shell Commands

- `%fs` → Interact with **DBFS (Databricks File System)**

```bash
%fs ls /FileStore/
```

```bash
%fs head /FileStore/data.txt
```

```bash
%fs mkdirs /FileStore/new_folder
```

```bash
%fs rm /FileStore/data.txt
```

- `%sh` → Run **Linux shell commands**

```bash
%sh
ls -l
```

```bash
%sh
echo "Hello Databricks"
```

## 🚀 Section 4: dbutils (Core Utilities)

- `dbutils` provides **programmatic control over environment**

### File Operations (`dbutils.fs`)
```python
dbutils.fs.ls("/FileStore/")
```

```python
dbutils.fs.cp("/FileStore/file1.txt", "/FileStore/file2.txt")
```

```python
dbutils.fs.mv("/FileStore/file1.txt", "/FileStore/archive/file1.txt")
```

```python
dbutils.fs.rm("/FileStore/file1.txt", True)
```

```python
dbutils.fs.head("/FileStore/file1.txt")
```

---

### Secrets (`dbutils.secrets`)
```python
dbutils.secrets.get(scope="my_scope", key="db_password")
```

---

### Notebook Control (`dbutils.notebook`)
```python
dbutils.notebook.exit("Success")
```

```python
dbutils.notebook.run("child_notebook", 60)
```

---

### Widgets (`dbutils.widgets`)
```python
dbutils.widgets.text("name", "default_value")
```

```python
name = dbutils.widgets.get("name")
print(name)
```

```python
dbutils.widgets.dropdown("env", "dev", ["dev", "prod"])
```
