###Extract Data from Json Order files
  1. Query file as JSON
  2. Query file as TEXT
  3. Create orders view in Bronze schema

###1. Query file as Json

```sql
select * from json.`/Volumes/ginzobox/landing/operational_data/orders`
```

###2. Query file using TEXT format

```sql
select * from text.`/Volumes/ginzobox/landing/operational_data/orders`
```

###3. Create views in bronze schema

```sql
create or replace view ginzobox.bronze.v_orders as select * from text.`/Volumes/ginzobox/landing/operational_data/orders`;
select * from ginzobox.bronze.v_orders
```
