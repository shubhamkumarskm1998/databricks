###Extract address data using read_files
  1. Demonstrate limitations of CSV file format
  2. Use read_files() function
  3. Create addresses view in bronze layer.

###1.Demonstrate limitations of CSV file format.

```sql
select * from csv.`/Volumes/ginzobox/landing/operational_data/addresses/*`
```

###2. Using read_files()

```sql
select * from read_files('/Volumes/ginzobox/landing/operational_data/addresses',
  format=>'csv',delimiter=>'\t',header=>true)
```

###3. Create address view in bronze layer.

```sql
create or replace view ginzobox.bronze.v_addresses as 
select * from read_files('/Volumes/ginzobox/landing/operational_data/addresses',
  format=>'csv',delimiter=>'\t',header=>true)
```
