###Set-up project env for Ginzobox data lakehouse

#### 1. Access the container

```sql
%fs ls 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/'
```

### 2. Create external location

```sql
%sql
Create external Location if not exists dea_course_ext_dl_ginzobox
URL 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/'
with (storage credential dea-course-ext-sc)
comment "DEA Ginzobox Project External Location"
```

###3. Create Ginzobox catalog

```sql
Create catalog if not exists ginzobox
managed location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/'
comment "This is the catalog for Ginzobox datalakehouse"
```

```sql
show catalogs;
```

###4. Create 4 schema's:
  1. landing
  2. bronze
  3. silver
  4. gold

```sql
select current_catalog();
```

```sql
Use catalog ginzobox;
```

```sql
select current_catalog();
```

```sql
create schema if not exists landing
  Managed location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/landing'
  comment 'This is the landing schema for Ginzobox datalakehouse'
```

```sql
show schemas;
```

```sql
create schema if not exists bronze
  Managed location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/bronze'
  comment 'This is the bronze schema for Ginzobox datalakehouse' ;
create schema if not exists silver
  Managed location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/silver' 
  comment 'This is the silver schema for Ginzobox datalakehouse';
create schema if not exists gold
  Managed location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/gold'
  comment 'This is the gold schema for Ginzobox datalakehouse';
```

```sql
show schemas;
```

###5. Create Volume

```sql
use catalog ginzobox;
use schema landing;

create external volume if not exists operational_data
  location 'abfss://ginzobox@deacourseextdlshubh.dfs.core.windows.net/landing/operational_data'
  comment 'This is the operational data volume for Ginzobox datalakehouse'
```

```sql
%fs ls /Volumes/ginzobox/landing/operational_data
```
