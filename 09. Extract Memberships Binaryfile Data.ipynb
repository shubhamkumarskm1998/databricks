###Extract data from memberships folder - Image files
  1. Query memberships files using binaryfile Format
  2. Create views in bronze schema

###1. Query memberships files using binaryfile Format

```sql
select * from binaryfile.`/Volumes/ginzobox/landing/operational_data/memberships/*/*.png`
```

###2. Create memberships views

```sql
create or replace view ginzobox.bronze.v_memberships as 
select * from binaryfile.`/Volumes/ginzobox/landing/operational_data/memberships/*/*.png`;
select * from ginzobox.bronze.v_memberships
```
